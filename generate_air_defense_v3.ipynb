```python
import numpy as np
import pandas as pd
import random
from datetime import datetime, timedelta

def generate_air_defense_data(samples=500):
    data = []
    
    # Possible values for categorical data
    ew_activity = ["None", "Jamming", "Spoofing", "Unknown"]
    weapon_status = ["Armed", "Unarmed", "Unknown"]
    object_behavior = ["Loitering", "Aggressive Maneuvers", "Direct Approach"]
    formation_type = ["Echelon", "Line Abreast", "V-Formation", "None"]
    country_codes = ["US", "RU", "CN", "FR", "UK", "DE", "IN", "IR", "PK", "KR"]
    time_of_day = ["Day", "Night", "Twilight"]
    terrain_types = ["Mountainous", "Flat", "Coastal"]
    solar_activity = ["Low", "Medium", "High"]
    
    start_time = datetime.now()
    
    for _ in range(samples):
        # Basic Detection & Tracking
        timestamp = start_time + timedelta(seconds=random.randint(0, 100000))
        latitude = round(np.random.uniform(-90, 90), 6)
        longitude = round(np.random.uniform(-180, 180), 6)
        vertical_speed = round(np.random.uniform(-50, 50), 2)
        acceleration = [round(np.random.uniform(-5, 5), 2) for _ in range(3)]
        bearing_change_rate = round(np.random.uniform(-10, 10), 2)
        altitude_change_rate = round(np.random.uniform(-20, 20), 2)
        signal_strength = round(np.random.uniform(-100, 0), 2)
        frequency = round(np.random.uniform(1e9, 10e9), 2)  # GHz range
        pri = round(np.random.uniform(1, 1000), 2)
        pulse_width = round(np.random.uniform(1, 100), 2)
        
        # Threat Assessment
        ew = random.choice(ew_activity)
        iff_code = random.randint(1000, 9999)
        flight_plan = {
            "ID": f"FP{random.randint(1000, 9999)}",
            "Departure": f"Airport_{random.randint(1, 50)}",
            "Arrival": f"Airport_{random.randint(1, 50)}",
            "Planned Altitude": round(np.random.uniform(1000, 40000), 2),
            "Planned Speed": round(np.random.uniform(200, 900), 2)
        }
        comm_signals = {
            "Frequency": round(np.random.uniform(100, 10000), 2),
            "Content": "Encrypted" if random.random() > 0.7 else "Standard",
            "Language": "Unknown" if random.random() > 0.8 else "English"
        }
        weapon_stat = random.choice(weapon_status)
        formation_count = random.randint(1, 6)
        formation_type_selected = random.choice(formation_type) if formation_count > 1 else "None"
        country = random.choice(country_codes)
        behavior = random.choice(object_behavior)
        
        # Environmental Data
        weather = {
            "Visibility": round(np.random.uniform(500, 50000), 2),
            "Wind Speed": round(np.random.uniform(0, 30), 2),
            "Cloud Cover": round(np.random.uniform(0, 100), 2)
        }
        tod = random.choice(time_of_day)
        terrain = random.choice(terrain_types)
        proximity = round(np.random.uniform(500, 100000), 2)
        air_traffic_density = random.randint(0, 100)
        solar = random.choice(solar_activity)
        
        # Sensor Fusion & Confidence
        sensor_reliability = round(np.random.uniform(50, 100), 2)
        confidence_detection = round(np.random.uniform(50, 100), 2)
        confidence_classification = round(np.random.uniform(50, 100), 2)
        num_sensors = random.randint(1, 5)
        sensor_id = f"SENSOR_{random.randint(1, 20)}"
        
        # Store data
        data.append([
            timestamp, latitude, longitude, vertical_speed, *acceleration,
            bearing_change_rate, altitude_change_rate, signal_strength, frequency, pri, pulse_width,
            ew, iff_code, flight_plan, comm_signals, weapon_stat, formation_count,
            formation_type_selected, country, behavior,
            weather, tod, terrain, proximity, air_traffic_density, solar,
            sensor_reliability, confidence_detection, confidence_classification,
            num_sensors, sensor_id
        ])
    
    # Create DataFrame
    columns = [
        "Time of Detection", "Latitude", "Longitude", "Vertical Speed", "Acceleration X", "Acceleration Y", "Acceleration Z",
        "Bearing Change Rate", "Altitude Change Rate", "Signal Strength", "Frequency", "PRI", "Pulse Width",
        "Electronic Warfare", "IFF Code", "Flight Plan", "Communication Signals", "Weapon Status", "Number of Objects in Formation",
        "Formation Type", "Country of Origin", "Object Behavior",
        "Weather Conditions", "Time of Day", "Geographic Terrain", "Proximity to Sensitive Locations",
        "Current Air Traffic Density", "Solar Activity",
        "Sensor Reliability", "Confidence Level (Detection)", "Confidence Level (Classification)",
        "Number of Sensors Detecting Object", "Sensor ID"
    ]
    
    df = pd.DataFrame(data, columns=columns)
    
    # Save to CSV
    df.to_csv("air_defense_v3.csv", index=False)
    print("Dataset 'air_defense_v3.csv' generated successfully!")
    
# Run the generator
generate_air_defense_data(500)
```
