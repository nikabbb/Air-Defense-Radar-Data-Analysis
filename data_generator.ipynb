```python
import numpy as np
import pandas as pd

#number of samples

num_samples = 1000

#object categories

object_type = ['Aircraft', 'Missile', 'Drone']

#generate random data

np.random.seed(42)

distance = np.random.uniform(10, 300, num_samples) #distance in km

speed = np.random.choice([np.random.uniform(200, 900), #aircraft speed
                          np.random.uniform(1000, 15000), #missile speed
                          np.random.uniform(50, 150)], #drone speed
                         num_samples) 

signal_strength = np.random.uniform(-10, 40, num_samples) #db
frequency_shift = np.random.uniform(1, 100, num_samples) #MHz

#assign labels based on speed range

labels = np.where(speed > 1000, 'Missile', np.where(speed > 200, 'Aircraft', 'Drone'))

df = pd.DataFrame({
    'Distance_km': distance,
    'Speed_mps': speed,
    'Signal_db': signal_strength,
    'Frequency_MHz': frequency_shift,
    'object_Type': labels
})

df.to_csv('radar_data.csv', index=False)

print(df.head())

```

```python
import pandas as pd
import numpy as np

# Define the number of samples
num_samples = 500  

# Define object types
object_types = ["Drone", "Missile", "Aircraft"]

# Set random seed for reproducibility
np.random.seed(42)

# Generate radar signal features with added noise
data = {
    "signal_strength": np.random.uniform(20, 100, num_samples) + np.random.normal(0, 15, num_samples),  # More noise
    "frequency": np.random.uniform(1, 5, num_samples) + np.random.normal(0, 0.7, num_samples),  # Higher variation
    "distance": np.random.uniform(500, 5000, num_samples) + np.random.normal(0, 300, num_samples),  # More random error
    "velocity": np.random.uniform(50, 1500, num_samples) + np.random.normal(0, 70, num_samples),  # Higher deviation
    "altitude": np.random.uniform(100, 15000, num_samples) + np.random.normal(0, 800, num_samples),  # More extreme values
    "Object_Type": np.random.choice(object_types, num_samples, p=[0.35, 0.3, 0.35])  # Slightly different class distribution
}

# Convert to DataFrame
df_noisy = pd.DataFrame(data)

# Introduce **intentional errors** (5% of dataset) - Simulate misclassifications
for _ in range(int(num_samples * 0.05)):  
    random_index = np.random.randint(0, num_samples)
    df_noisy.loc[random_index, "Object_Type"] = np.random.choice(object_types)

# Save to CSV
df_noisy.to_csv("noisy_radar_data.csv", index=False)

print("Noisy dataset saved as noisy_radar_data.csv")
```
