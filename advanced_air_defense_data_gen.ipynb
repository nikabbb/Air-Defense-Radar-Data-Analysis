```python
# Importing necessary libraries
import numpy as np
import pandas as pd
import random
import faker
from sklearn.preprocessing import LabelEncoder

# Initialize Faker for generating fake data
fake = faker.Faker()

# Number of data points
n = 1000

# Function to generate random drone/aircraft/missile data
def generate_data(n):
    # Define possible classes and categories
    radar_types = ['Long-Range', 'Medium-Range', 'Short-Range']
    detection_types = ['Active', 'Passive', 'Hybrid']
    missile_types = ['Cruise Missile', 'Ballistic Missile', 'Anti-Air Missile', 'Surface-to-Air Missile']
    aircraft_types = ['Fighter Jet', 'Bombers', 'Drones', 'Helicopter', 'Passenger Aircraft']
    drone_types = ['Quadcopter', 'Fixed-Wing', 'Hybrid Drone']
    threat_levels = ['High', 'Medium', 'Low']
    
    # Initialize data
    data = []
    
    for _ in range(n):
        radar_type = random.choice(radar_types)
        detection_type = random.choice(detection_types)
        missile_type = random.choice(missile_types)
        aircraft_type = random.choice(aircraft_types)
        drone_type = random.choice(drone_types)
        threat_level = random.choice(threat_levels)
        
        # Generate additional attributes
        altitude = round(np.random.uniform(500, 20000), 2)  # in meters
        speed = round(np.random.uniform(200, 1500), 2)  # in km/h
        distance_from_target = round(np.random.uniform(50, 500), 2)  # in km
        heading = round(np.random.uniform(0, 360), 2)  # degrees
        radar_signature = round(np.random.uniform(0, 1), 2)  # radar cross section (RCS)
        
        # Randomly simulate if threat is detected or not (for binary classification purpose)
        detected = np.random.choice([0, 1], p=[0.2, 0.8])  # 80% chance of detection
        
        # Simulate multi-class target (drone, aircraft, missile)
        if detected == 1:
            target_class = random.choice([missile_type, aircraft_type, drone_type])
        else:
            target_class = "None"
        
        data.append([radar_type, detection_type, missile_type, aircraft_type, drone_type, 
                     threat_level, altitude, speed, distance_from_target, heading, radar_signature, 
                     detected, target_class])
    
    # Create DataFrame
    columns = ['Radar Type', 'Detection Type', 'Missile Type', 'Aircraft Type', 'Drone Type', 
               'Threat Level', 'Altitude', 'Speed', 'Distance From Target', 'Heading', 
               'Radar Signature', 'Detection Status', 'Target Class']
    
    df = pd.DataFrame(data, columns=columns)
    return df

# Generate the dataset
df = generate_data(n)

# Saving to a CSV file
df.to_csv('air_defense_advanced_data.csv', index=False)

# Display first few rows of the generated data
df.head()
```
