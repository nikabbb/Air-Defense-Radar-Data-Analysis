```python
import pandas as pd

df = pd.read_csv('air_defense_advanced_data.csv')

print(df.info())

print(df.head())

print(df.describe())

```

```python
#missing values
print(df.isnull().sum())
```

```python
#visualization of target variable distribution
import matplotlib.pyplot as plt
import seaborn as sns

#count plot for target class - the label we want to predict
plt.figure(figsize=(8, 6))
sns.countplot(data=df, x='Target Class', palette='Set2')
plt.title('Distribution of Target Class')
plt.show()
```

```python
#selecting only numerical columns
numerical_df = df.select_dtypes(include=['float64', 'int64'])

#heatmap to explore correlations between numerical features:
plt.figure(figsize=(10, 8))
sns.heatmap(numerical_df.corr(), annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Correlation Heatmap of Numerical Features')
plt.show()
```
